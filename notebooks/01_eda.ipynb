```python
duckdb.sql("""
    SELECT order_status, COUNT(*) AS cnt,
           ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) AS pct
    FROM 'D:/olist-user-dashboard/data/raw/olist_orders_dataset.csv'
    GROUP BY order_status
    ORDER BY cnt DESC
""").show()
```

```python
import duckdb

tables = {
    "客户表": "olist_customers_dataset.csv",
    "订单表": "olist_orders_dataset.csv",
    "订单商品表": "olist_order_items_dataset.csv",
    "支付表": "olist_order_payments_dataset.csv",
    "评价表": "olist_order_reviews_dataset.csv",
    "商品表": "olist_products_dataset.csv",
    "卖家表": "olist_sellers_dataset.csv",
    "地理位置表": "olist_geolocation_dataset.csv",
    "品类翻译表": "product_category_name_translation.csv",
}

for 名称, 文件 in tables.items():
    cnt = duckdb.sql(f"SELECT COUNT(*) FROM 'D:/olist-user-dashboard/data/raw/{文件}'").fetchone()[0]
    print(f"{名称:12s} {cnt:>10,} 行")
```

```python
duckdb.sql("""
    SELECT * FROM 'D:/olist-user-dashboard/data/raw/olist_orders_dataset.csv' LIMIT 5
""").show()
```

```python
duckdb.sql("""
    SELECT
        order_status AS 订单状态,
        COUNT(*) AS 订单数,
        ROUND(COUNT(*) * 100.0 / SUM(COUNT(*)) OVER (), 2) AS 占比百分比
    FROM 'D:/olist-user-dashboard/data/raw/olist_orders_dataset.csv'
    GROUP BY order_status
    ORDER BY 订单数 DESC
""").show()
```

```python
duckdb.sql("""
    SELECT
        COUNT(*) AS 总行数,
        COUNT(DISTINCT customer_id) AS 订单级客户数,
        COUNT(DISTINCT customer_unique_id) AS 真实用户数
    FROM 'D:/olist-user-dashboard/data/raw/olist_customers_dataset.csv'
""").show()
```

```python
duckdb.sql("""
    SELECT
        MIN(order_purchase_timestamp) AS 最早购买时间,
        MAX(order_purchase_timestamp) AS 最晚购买时间
    FROM 'D:/olist-user-dashboard/data/raw/olist_orders_dataset.csv'
""").show()
```

```python
duckdb.sql("""
    SELECT
        product_category_name AS 葡萄牙语品类名,
        product_category_name_english AS 英文品类名
    FROM 'D:/olist-user-dashboard/data/raw/product_category_name_translation.csv'
    LIMIT 10
""").show()
```
